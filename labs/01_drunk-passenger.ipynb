# Lab 01a: The Drunk Passenger problem

Let's say that we have an airplane with 100 seats, and every seat is booked. Each of the 100 passengers has a ticket with a seat assignment. For simplicity, let's assume Passenger 1 is in Seat 1, Passenger 2 is in Seat 2, and so forth.

But there's a wrinkle: Passenger 1 isn't a great traveller, so to try and calm their nerves they've had a little bit to drink and are slightly more tipsy than they should be. When Passenger 1 boards, they forget what seat they're assigned and sit in any random seat instead.

For each passenger who boards after them, if their assigned seat is available, they will sit in it. If their assigned seat is already taken, they will also choose a new, random seat.

**Write a program that boards all 100 passengers and returns the final seat assignment for Passenger 100. Does Passenger 100 get to keep their original seat or not?**

```python
import random

# We are using a list comprehension to create a list of passengers,
# numbered 1 through 100. This is equivalent to using a for loop
# to append items to the list. Note that the `end` argument in our
# range function is exclusive — 101 will not be included in our list
# of passengers
passengers = [passenger for passenger in range(1, 101)]
available_seats = [seat for seat in range(1, 101)]

# Choose a random seat number for Passenger 1. The `sample()` function
# lets us provide a list of valid values to choose from, and will
# randomly select `k` elements from that list. In our case, we are
# choosing only 1 seat at a time.
#
# Note that I'm adding a list slice to the end of the function — the [0] —
# this is because `sample()` returns a list of randomly selected values,
# but since we're only returning 1 value at a time, it's easier for us
# to treat it as an integer. `[0]` means that we want to extract only the
# zero-th element of the list, and are ignoring any other items
random_seat = random.sample(population=available_seats, k=1)[0]

# Now that we've chosen a random seat for our passenger, we need to remove
# it from the list of still-available seats. We will use the `remove()`
# method to do this
available_seats.remove(random_seat)

# Finally, we will re-run this entire process for the other 99 passengers
#
# We can structure our for loop in one of two ways: either we can operate
# on list indices, or directly on the list of passengers. For illustration,
# a for loop with list indices is included in the comments below. The
# uncommented code operates directly on the passenger list
#
# for i in range(1, 100):
#     if passengers[i] in available_seats:
#         available_seats.remove(passengers[i])
#     else:
#         available_seats.remove(random.sample(available_seats, 1)[0])

for passenger in passengers[1:99]:
    if passenger in available_seats:
        available_seats.remove(passenger)
    else:
        random_seat = random.sample(population=available_seats, k=1)[0]
        available_seats.remove(random_seat)

# Finally, we will evaluate whether passenger 100 (list index 99) has their
# seat still available or not
print('🛩️' if passengers[99] in available_seats else '😭')
```

# Lab 01b: Return of the Drunk Passenger

Great! So we've figured out that our poor final passenger to board got to keep their seat. Or maybe not. Really, who can say. Pseudorandom number generators, am I right?

So let's extend our original program a bit. Instead of just boarding everyone once, let's do it 10,000 times!

**Write a program that simulates the boarding process from Lab 01a 10,000 times. Keep track of how many times Passenger 100 is able to keep their assigned seat. What is the probability of Passenger 100 keeping their assigned seat?**

```python
import random


keeps_seat = []

for _ in range(10000):
    passengers = [passenger for passenger in range(1, 101)]
    available_seats = [seat for seat in range(1, 101)]

    random_seat = random.sample(population=available_seats, k=1)[0]

    available_seats.remove(random_seat)

    for passenger in passengers[1:99]:
        if passenger in available_seats:
            available_seats.remove(passenger)
        else:
            random_seat = random.sample(population=available_seats, k=1)[0]
            available_seats.remove(random_seat)

    keeps_seat.append(passengers[99] in available_seats)

print(sum(keeps_seat) / len(keeps_seat))
```
