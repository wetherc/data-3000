# Reordering Supply Stacks

## Background

The expedition can depart as soon as the final supplies have been unloaded from the ships. Supplies are stored in stacks of marked crates, but because the needed supplies are buried under many other crates, the crates need to be rearranged.

The ship has a giant cargo crane capable of moving crates between stacks. To ensure none of the crates get crushed or fall over, the crane operator will rearrange them in a series of carefully-planned steps. After the crates are rearranged, the desired crates will be at the top of each stack.

The Elves don't want to interrupt the crane operator during this delicate procedure, but they forgot to ask her which crate will end up where, and they want to be ready to unload them as soon as possible so they can embark.

They do, however, have a drawing of the starting stacks of crates and the rearrangement procedure (your puzzle input). For example:

```
    [D]    
[N] [C]    
[Z] [M] [P]
 1   2   3

move 1 from 2 to 1
move 3 from 1 to 3
move 2 from 2 to 1
move 1 from 1 to 2
```

In this example, there are three stacks of crates. Stack 1 contains two crates: crate Z is on the bottom, and crate N is on top. Stack 2 contains three crates; from bottom to top, they are crates M, C, and D. Finally, stack 3 contains a single crate, P.

Then, the rearrangement procedure is given. In each step of the procedure, a quantity of crates is moved from one stack to a different stack. In the first step of the above rearrangement procedure, one crate is moved from stack 2 to stack 1, resulting in this configuration:

```
[D]        
[N] [C]    
[Z] [M] [P]
 1   2   3
```

In the second step, three crates are moved from stack 1 to stack 3. Crates are moved one at a time, so the first crate to be moved (D) ends up below the second and third crates:

```
        [Z]
        [N]
    [C] [D]
    [M] [P]
 1   2   3
```

Then, both crates are moved from stack 2 to stack 1. Again, because crates are moved one at a time, crate C ends up below crate M:

```
        [Z]
        [N]
[M]     [D]
[C]     [P]
 1   2   3
```

Finally, one crate is moved from stack 1 to stack 2:

```

        [Z]
        [N]
        [D]
[C] [M] [P]
 1   2   3
```

The Elves just need to know which crate will end up on top of each stack; in this example, the top crates are C in stack 1, M in stack 2, and Z in stack 3, so you should combine these together and give the Elves the message CMZ.

After the rearrangement procedure completes, what crate ends up on top of each stack?

## Solution

````python
# We will start by reading in the file, one line at a time, using the
# readlines() method. As a reminder, the
# 
#   with open('filename', 'r') as f
# 
# line is called a context manager in Python: in this case, it's an
# easy way to handle a lot of the nitty gritty of opening / reading /closing
# files.
with open('lab_02.txt', 'r') as f:
    data = [line.strip() for line in f.readlines()]

for line in data[:15]:
    print(line)
````

````python
# Since our input is just a list of strings that will need to be parsed,
# we will create a couple of variables here to contain the important
# pieces of information we parse out
crates = []
instructions = []

for line in data:
    # We know that any crate in our input will always look like
    # 
    #   [ID]
    # 
    # so we can make a conditional statement to look for the opening
    # square bracket anywhere in the line. If it exists, we know that we're
    # looking at a row of crates!
    if line and '[' in line:
        # The entire row of crates actually has a lot of information that we
        # don't need, so here what we will do is look at the line of text
        # one character at a time, and extract the value if that character's
        # position in the string is evenly divisible by 4. This is because we
        # know that the ID of a crate will appear exactly every 4th character
        # in the line. The `enumerate()` function is what lets us move through
        # the string character-by-character, and keeps track of the current
        # character's index in the string using the `idx` variable that we
        # create
        crate = [
            ((idx - 1) // 4, elem) for idx, elem in enumerate(line)
            if (idx - 1) % 4 == 0
        ]
        crates.append(crate)
    # Similar to above, we know that any line of instruction text will always
    # begin with the word "move", so it's easy for us to just check whether
    # a line begins with the letter "m" to identify that it's an instruction
    elif line and line[0] == 'm':
        # We're converting the line of instructions
        # 
        #   move X from Y to Z
        # 
        # into a list of integers of the form [X, Y, Z] so we always know
        # the number of crates to move, from where, and to where
        instructions.append([
            int(value) for value in line.split()
            if value.isdigit()
        ])
````

````python
# Although moving veritcally-stacked crates makes logical sense to us,
# to Python, it's actually a lot easier to think of things in terms of
# rows of crates, with the topmost crate being the furthest to the right.
# Doing this lets us just pop an element off the end of one list and append
# it to the end of another without much fuss

# This should really be hardcoded, but we'll let it slide 🙂
rotated_crates = [[] for _ in range(9)]
for row_num, row_values in enumerate(crates):
    for crate in row_values:
        if crate[1] != ' ':
            rotated_crates[crate[0]].append(crate[1])

for position, line in enumerate(rotated_crates):
    rotated_crates[position].reverse()

for line in rotated_crates:
    print(line)
````

````python
for instruction in instructions:
    # Now for each crate that needs to be moved, we can easily lift that
    # crate off the top of its starting stack using the .pop() method
    # and move it to the top of its destination stack with the .append() method
    for crate_number in range(instruction[0]):
        _source_stack = instruction[1] - 1
        _destination_stack = instruction[2] - 1

        _crate = rotated_crates[_source_stack].pop()
        rotated_crates[_destination_stack].append(_crate)

for row in rotated_crates:
    print(row)
````

In this example, the CrateMover 9001 has put the crates in a totally different order: MCD.

Before the rearrangement process finishes, update your simulation so that the Elves know where they should stand to be ready to unload the final supplies. After the rearrangement procedure completes, what crate ends up on top of each stack?

````python
with open('lab_02.txt', 'r') as f:
    data = [line.strip() for line in f.readlines()]

crates = []
instructions = []

for line in data:
    if line and '[' in line:
        crate = [
            ((idx - 1) // 4, elem) for idx, elem in enumerate(line)
            if (idx - 1) % 4 == 0
        ]
        crates.append(crate)
    elif line and line[0] == 'm':
        instructions.append([
            int(value) for value in line.split()
            if value.isdigit()
        ])
````

````python

rotated_crates = [[] for _ in range(9)]
for row_num, row_values in enumerate(crates):
    for crate in row_values:
        if crate[1] != ' ':
            rotated_crates[crate[0]].append(crate[1])

for position, line in enumerate(rotated_crates):
    rotated_crates[position].reverse()
````

````python
for line in rotated_crates:
    print(line)
````

````python
for instruction in instructions:
    stack_len = len(rotated_crates[instruction[1] - 1])
    _crates_to_move = rotated_crates[instruction[1] - 1][-instruction[0]:]

    del rotated_crates[instruction[1] - 1][-instruction[0]:]

    rotated_crates[instruction[2] - 1].extend(_crates_to_move)
````
